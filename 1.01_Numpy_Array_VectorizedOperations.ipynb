**Numpy 배열 - 벡터화연산**

```python
# NumPy 배열의 ‘벡터화 연산’cf, 파이썬 반복문
# 실행 시간 비교
```

```python
import numpy as np
import time
```

```python
# 데이터 생성
data_list = list(range(1, 1000001))  # 1부터 1,000,000까지의 리스트 생성
data_array = np.array(data_list)       # 같은 데이터를 가진 NumPy 배열 생성
```

```python
# 비교 - 1
# 반목문을 사용한 연산
start_time = time.time()

result_list = []
# 파이썬 반목문, 요소를 하나씩 꺼내 계산하고 결과를 리스트에 추가
for x in data_list:
    result_list.append(x * 2)

end_time = time.time()
loop_duration = end_time - start_time

print(f"루프를 사용한 연산 시간: {loop_duration:.6f}초")
```

```python
# 비교 - 2
# 벡터화 연산 (Vectorized Operations)
start_time = time.time()

# NumPy 벡터화 연산, 배열 전체에 적용할 계산을 한 번에 표현
result_array = data_array * 2    # 모든 요소에 *2 연산

end_time = time.time()
vectorized_duration = end_time - start_time

print(f"Vectorized Operations 연산 시간: {vectorized_duration:.6f}초")
```

```python
# 결과 비교
print(f"Vectorized Operations 연산 시간 빠르기 : {loop_duration / vectorized_duration:.2f}배 빠름")
```

```python
# 파이썬 내장 딕셔너리

student = {
    "회원 ID": "S001",    # str
    "학습 시간": 12.5,    # float
    "제출 횟수": 4,       # int
    "만족도": "높음",     # str
    "수료 여부": True     # bool
}

print(student["학습 시간"])  # 키로 접근 → 12.5
```

```python
# 리스트 <-- 딕셔너리들을 담음

students = [
    {"회원 ID": "S001", "학습 시간": 12.5, "제출 횟수": 4,
     "만족도": "높음", "수료 여부": True},

    {"회원 ID": "S002", "학습 시간": 8.0, "제출 횟수": 2,
     "만족도": "보통", "수료 여부": False},

    {"회원 ID": "S003", "학습 시간": 15.0, "제출 횟수": 5,
     "만족도": "높음", "수료 여부": True}
]

print(students[0]["회원 ID"])  # 첫 번째 회원의 ID -> S001
```

```python
# 튜플, 집합

# 튜플: 한 회원의 정보를 순서대로 묶기
student_tuple = ("S001", 12.5, 4, "높음", True)

# 집합: 데이터에 등장한 만족도 값을 중복 없이 담기
satisfaction_set = {"높음", "보통", "높음"} # <-- 실제로는 "높음", "보통" 두 요소만 보관, 순서 X
```
